### Agentic AI Systems Vs AI Agents

- AI Agents : It refers to software programs designed to perform specific tasks without human intervention, with a degree  autonomy.
Example: LLMs are basically trained on data and if you ask a simple LLM a question on which it is trained on will give the correct output. But if we ask a question like "What's the temperature in Banglore?" it won't be able to answer, it would require a external data source for this LLM to answer this question. This External Data Source can be any API or Database. The LLM itself directs the query to the external data source which shows the degree of autonomy and thus we can say it as an Basic AI Agent.

- Agentic AI Systems : It is a framework where multiple AI Agents can collaborate and make decisions independently to achieve a larger goal.
Example: It is a broader framework and not just one AI Agent it has multiple AI Agents which can communicate between each other to solve or reach a specific goal. 
We need to generate a blog from the youtube video then firstly we need to generate transcript from the yt video which will be done by some AI Agent and then that transcript will be sent to another AI Agent to generate the blog title, description,etc. Here the first AI Agent will provide the transcript to the second AI Agent which will take the trasncript as the input and give the blog as the output and hence they both communicate with each other.
