### Pydantic Basics : Creating and Using Models

Pydantic modles are the foundation of data validation in Python. They use Python type annotations to define the structure and validate data at runtime. Here's a detailed exploration of basic model creation with several examples.


```python
from pydantic import BaseModel
```

```python
from dataclasses import dataclass

@dataclass
class Person1():
    name: str
    age: int
    city: str

person = Person1(name="Piyush", age=22, city=45)
print(person)
```

### 2. Model with Optional Fields
Add optional fields using Python's Optional Type:

```python
from typing import Optional
class Employee(BaseModel):
    name: str
    age: int
    city: str
    salary: Optional[float] = None  # Optional with default value None
    is_active: Optional[bool] = True # Optional with default value True

```

```python
# Examples with and without optional fields

employee1 = Employee(name="Alice", age=30, city="New York")
print(employee1)
```

```python
employee2 = Employee(name="Bob", age=25, city="Los Angeles", salary=50000.0, is_active=False)
print(employee2)
```

Definition:
- Optional[type] : Indicates that the field can be None

- Default Value (=None or =True): Makes the field optional
 
- Required fields still must be provided

- Pydantic evaluates types even for the optional fields when values are provided 

```python
from pydantic import BaseModel
from typing import List
class Classroom(BaseModel):
    name: str
    students: List[str]  # List of String i.e Student Names
    capacity: int
```

```python
#Create a classroom
classroom = Classroom(name="Math 101", students=("Alice", "Bob", "Charlie"), capacity=30)
print(classroom)
```

### 4. Model with Nested Models
Create complex structures with nested models:


```python
from pydantic import BaseModel

class Address(BaseModel):
    street: str
    city: str
    zip_code: int  

class Customer(BaseModel):
    name: str
    email: str
    address: Address # Nested Model

# Create a customer with nested address
customer = Customer(
    name="John Doe",
    email="jhone@123.com",
    address=Address(street="123 Main St", city="Anytown", zip_code=12345),
)
print(customer)
```

### Pydantic Fields: Customization and Constraints 

The field function in Pydantic enhances model fields beyond basic type hints by allowing you to specify validation rules, default values, aliases, and more. Here's more comprehensive tutorial with examples.


```python
from pydantic import BaseModel, Field
class Item(BaseModel):
    name: str = Field(min_length=2, description="The name of the item", max_length=50) # Validation for string length
    price: float = Field(gt=0, description="The price must be greater than zero", le=1000) # Validation for numeric range
    quantity: int = Field(ge=0, description="The quantity must be at least 1") # Validation for non-negative integers

# Create an item with valid data
item = Item(name="Laptop", price=999.55, quantity=10)
print(item)
```

```python
from pydantic import BaseModel, Field
class User(BaseModel):
    username: str = Field(..., description="Username")
    email: str = Field(default_factory=lambda:"user@example.com", description="Must be a valid email address")
    age: int = Field(default=18, le=100, description="User Age, Default is 18, must be less than or equal to 100")

# Create a user with only username
user1 = User(username="john_doe")
print(user1)

# Create a user with all fields
user2 = User(username="jane_doe", email="jane@gmail.com", age=25)
print(user2)
```
